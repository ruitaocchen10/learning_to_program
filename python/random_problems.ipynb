```python
# Two Sum

# Given a list of integers nums and an integer target, return the indices of the two numbers that add up to target. Each input has exactly onese the same element twice.

# two_sum([2, 7, 11, 15], 9)  # -> [0,
# two_sum([3, 2, 4], 6)       # -> [1, 2]
# two_sum([3, 3], 6)          # -> [0,

# Start with whatever approach comes tobrute force. Then try to bring it downfrom O(n²) to O(n).

class Solution:
    def twosum(self, nums, target):
        for i in range(len(nums)):
            for k in range(i + 1, len(nums)):
                if nums[i] + nums[k] == target:
                    return [i, k]

        return False

nums = [3, 2, 4]
target = 6
solution = Solution()
answer = solution.twosum(nums, target)
print(answer)
```

```python
# Two Sum

# Given a list of integers nums and an integer target, return the indices of the two numbers that add up to target. Each input has exactly onese the same element twice.

# Start with whatever approach comes tobrute force. Then try to bring it downfrom O(n²) to O(n).

class Solution:
    def twosum(self, nums, target):
        seen = {}

        for i, num in enumerate(nums):
            sum = target - num
            if sum in seen:
                return [seen[sum], i]
            else:
                seen[num] = i

        return False


nums = [2, 7, 11, 15]
target = 9
solution = Solution()
answer = solution.twosum(nums, target)
print(answer)
```
