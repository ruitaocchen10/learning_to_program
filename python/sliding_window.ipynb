```python
# You are given an integer array prices where prices[i] is the price of NeetCoin on the ith day.

class Solution:
    def maxProfit(self, prices):
        left = 0
        right = 1
        output = 0

        while right < len(prices):
            if prices[right] < prices[left]:
                left = right
                right += 1
            else:
                temp = prices[right] - prices[left]
                if temp > output:
                    output = temp
                    right += 1
                else:
                    right += 1

        return output
    
prices = [7,1,5,3,6,4]
solution = Solution()
answer = solution.maxProfit(prices)
print(answer)
```

```python
# Given a string s, find the length of the longest substring without duplicate characters.

# A substring is a contiguous sequence of characters within a string.

class Solution:
    def lengthOfLongestSubstring(self, s):
        s_list = list(s)

        left = 0
        right = 1

        val = 0

        while right < len(s_list):
            range_list = s_list[left:right + 1]
        
            if s_list[right] not in range_list:
                right += 1
            elif s_list[right] not in range_list and right - 1 == left:
                right += 1
                left += 1
            else:
                left += 1

            if len(range_list) > val:
                val = len(range_list)

        return val


s = "zxyzxyz"
solution = Solution()
answer = solution.lengthOfLongestSubstring(s)
print(answer)

```
