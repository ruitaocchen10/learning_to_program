```python
# Given an integer array nums, return true if any value appears more than once in the array, otherwise return false.

class Solution:
    def hasDuplicate(self, nums):
        values = []
        for i in nums:
            if i not in values:
                values.append(i)
            else:
                return True
        
        return False


nums = [1,5,4,3,6]
solution = Solution()
answer = solution.hasDuplicate(nums)
print(answer)
```

```python
""" Given an array of integers nums, return the length of the longest consecutive sequence of elements that can be formed. """

class Solution:
    def longestConsecutive(self, nums):
        starters = []

        for n in nums:
            if n-1 not in nums:
                starters.append(n)

        count = 0

        for n in starters:
            nxt_val = n
            temp_count = 0
            while nxt_val in nums:
                nxt_val += 1
                if temp_count >= count:
                    count += 1
                    temp_count += 1
                else:
                    temp_count += 1

        return count
    
nums = [2,20,4,10,3,4,5]
solution = Solution()
answer = solution.longestConsecutive(nums) 
print(answer)
```

```python
""" You are given an array of distinct integers nums, sorted in ascending order, and an integer target.

Implement a function to search for target within nums. If it exists, then return its index, otherwise, return -1. """

class Solution:
    def search(self, nums, target):
        if nums[0] > target:
            return -1
        
        left = 0
        right = len(nums) - 1

        while left <= right:
            mv = left + ((right - left) // 2)
            if nums[mv] < target:
                left = mv + 1
            elif nums[mv] > target:
                right = mv -  1
            elif nums[mv] == target:
                return mv
            
        return -1

nums = [-1,0,2,4,6,8]
target = 4
solution = Solution()
answer = solution.search(nums, target)
print(answer)
        
```

```python
class Solution:
    def topKFrequent(self, nums, k):
        map = {}
        result = []

        for num in nums:
            if num not in map:
                map[num] = 1
            else:
                map[num] += 1

        while k > 0:
            max_val = max(map, key=max.get)
            result.append(max_val)
            map.pop(max_val)
            k -= 1

        return result

nums = [1,1,1,1,2,2,3]
k = 2
solution = Solution()
answer = solution.topKFrequent(nums, k)
print(answer)
```
