```python
""" Given two strings needle and haystack, return the index 
of the first occurrence of needle in haystack, or -1 if needle 
is not part of haystack. """


class Solution:
    def strStr(self, haystack, needle):
        hay_list = list(haystack)
        needle_list = list(needle)

        n = 0
        h = 0

        while h < len(hay_list):
            if hay_list[h] == needle_list[n]:
                h += 1
                n += 1
                if n == len(needle_list):
                    return h - n
            else:
                h = (h - n) + 1
                n = 0

        return -1

haystack = "leetcode"
needle = "leeto"
solution = Solution()
answer = solution.strStr(haystack, needle)
print(answer)

        
```

```python
""" Given an integer array nums and an integer val, remove all occurrences of val in nums in-place. The order of the elements may be changed. Then return the number of elements in nums which are not equal to val.

Consider the number of elements in nums which are not equal to val be k, to get accepted, you need to do the following things: """

class Solution:
    def removeElement(self, nums, val):
        index = 0
        for i in range(len(nums)):
            if nums[i] != val:
                nums[index] = nums[i]
                index += 1
        return nums
    
nums = [0,1,2,2,3,4,4,2]
val = 4
solution = Solution()
answer = solution.removeElement(nums, val)
print(answer)
```

```python
""" Given a sorted array of distinct integers and a target value, return the index if the target is found. If not, return the index where it would be if it were inserted in order.

You must write an algorithm with O(log n) runtime complexity. """

class Solution:
    def searchInsert(self, nums, target):
        left = 0
        right = len(nums) - 1
        length = (right - left) // 2
        
        while length > 0:
            index = length + left
            if nums[index] < target:
                left = index
            elif nums[index] > target:
                right = index
            elif nums[index] == target:
                return index

        return index

nums = [1,3,5,6,8,9,10]
target = 7
solution = Solution()
answer = solution.searchInsert(nums, target)
print(answer)
```

```python
""" You are given a large integer represented as an integer array digits, where each digits[i] is the ith digit of the integer. The digits are ordered from most significant to least significant in left-to-right order. The large integer does not contain any leading 0's.

Increment the large integer by one and return the resulting array of digits.

 """

class Solution:
    def plusOne(self, digits):
        num = int("".join(map(str, digits)))
        num += 1

        ans = str(num)

        answer = [int(d) for d in ans]

        return answer

        

digits = [1,2,3]
solution = Solution()
answer = solution.plusOne(digits)
print(answer)
```

```python
""" You are given a large integer represented as an integer array digits, where each digits[i] is the ith digit of the integer. The digits are ordered from most significant to least significant in left-to-right order. The large integer does not contain any leading 0's.

Increment the large integer by one and return the resulting array of digits.

 """

class Solution:
    def plusOne(self, digits):
        right = len(digits) - 1

        digits[right] += 1

        while right != 0:
            if digits[right] == 10:
                digits[right] = 0
                right -= 1
                digits[right] += 1
            else:
                break

        if digits[right] == 10:
            digits[right] = 0
            digits.insert(0, 1)

        return digits

digits = [9]
solution = Solution()
answer = solution.plusOne(digits)
print(answer)
```
